```python
import requests
from bs4 import BeautifulSoup

api_url = "https://theamazingworldofgumball.fandom.com/api.php"

params = {
    "action": "parse",
    "page": "The Amazing World of Gumball Wiki",
    "prop": "text",
    "format": "json",
    "formatversion": "2"
}

headers = {
    "User-Agent": "Mozilla/5.0"
}

response = requests.get(
    api_url,
    params=params,
    headers=headers
)

print(response.status_code)
```

```python
data = response.json()

html = data["parse"]["text"]

soup = BeautifulSoup(
    html,
    "html.parser"
)

print("Berhasil mendapatkan halaman!")
```

```python
headings = []

for heading in soup.select("h2, h3"):
    text = heading.get_text(" ", strip=True)

    if text:
        headings.append(text)
```

```python
for heading in headings:
    print(heading)
```

```python
from urllib.parse import urljoin
```

```python
base_url = "https://theamazingworldofgumball.fandom.com"

links = []

for a in soup.find_all("a", href=True):

    text = a.get_text(" ", strip=True)
    href = a["href"]

    if href.startswith("/wiki/") and text:

        full_url = urljoin(base_url, href)

        links.append({
            "Nama": text,
            "URL": full_url
        })
```

```python
df = pd.DataFrame(links)

df.head(20)
```

```python
df = df.drop_duplicates()

df = df.reset_index(drop=True)

df
```

```python
print("Jumlah link:", len(df))
```

```python
df.to_csv(
    "gumball_wiki.csv",
    index=False
)
```
